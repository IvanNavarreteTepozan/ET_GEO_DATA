```python
import pandas as pd


def intervalo_a_cols(intervalo):
    # Si el valor es nulo o vacío
    if pd.isna(intervalo) or str(intervalo).strip() == "":
        return (-1)
    
    intervalo = str(intervalo).strip("[]() ")
    
    try:
        # Intentar dividir por coma
        if "," in intervalo:
            a, b = intervalo.split(",")
        # Intentar dividir por guion
        elif "-" in intervalo:
            a, b = intervalo.split("-")
        else:
            # Solo un valor → regresa -100 en ambos
            return (-1)
        
        return (float(a)/200 + float(b)/200)
    
    except ValueError:
        # Si falla el split o la conversión
        return (-1)

# Crear DataFrame

def construir_clave_mun(row):
    text=str(row["CVE_MUN"])
    mun = text[-3:]
    
    return mun


def construir_clave_loc(row):
    text=str(row["CVE_LOC"])
    loc= text[-3:]
    
    return loc
```

```python
df.dropna(inplace=True)
df['rango_pobreza']=df.rango_pobreza.map(intervalo_a_cols)
```

```python
df = df[df["rango_pobreza"] != -1] #valores nulos o faltantes
```

```python
import numpy as np

# 1. Definimos la lista de columnas que vamos a evaluar
cols_ing = ['ing_1', 'ing_2', 'ing_3', 'ing_4', 'ing_5', 'ing_6']

# 2. Reemplazamos los 0 por NaN, y calculamos el promedio por fila (axis=1)
df['promedio_ing_mensual'] = df[cols_ing].replace(0, np.nan).mean(axis=1)

# 3. (Opcional) Si un hogar tiene ceros en TODOS los meses, el resultado será NaN. 
# Si prefieres que en esos casos vuelva a decir 0 en lugar de NaN, descomenta la siguiente línea:
# df['promedio_ing_mensual'] = df['promedio_ing_mensual'].fillna(0)

# Mostramos el resultado para comprobar
df[['ing_1', 'ing_2', 'ing_3', 'ing_4', 'ing_5', 'ing_6', 'promedio_ing_mensual']]
```

```python
import pandas as pd

# Asegurarnos de que estim_pago sea numérico por si hay espacios o errores de formato
df['estim_pago'] = pd.to_numeric(df['estim_pago'], errors='coerce')

# ---------------------------------------------------------
# PARTE A: Clasificación por ingresos (Clase Baja, Media, Alta)
# ---------------------------------------------------------

# ---------------------------------------------------------
# PARTE B: Conteo de estrato socioeconómico y estadística de estim_pago
# ---------------------------------------------------------

# 1. Contar el número de hogares por cada est_socio
df_est_socio = df.groupby('ubica_geo')['est_socio'].value_counts().unstack(fill_value=0)
df_est_socio.columns.name = None
# Renombramos para que sea fácil de identificar (suponiendo estratos del 1 al 4)
df_est_socio.columns = [f'est_socio_{int(col)}' for col in df_est_socio.columns]

# 2. Calcular media y desviación estándar de estim_pago
# Usamos .agg() para calcular múltiples estadísticas a la vez
df_estim_pago = df.groupby('ubica_geo')['promedio_ing_mensual'].agg(['mean'])

df_estim_pago['EstimPagoPromedio']=df_estim_pago['mean']
df_estim_pago=df_estim_pago['EstimPagoPromedio']
# ---------------------------------------------------------

df_final = pd.concat([ df_est_socio, df_estim_pago], axis=1).reset_index()


df_final = df_final.rename(columns={
    'est_socio_1': 'cantidadHogares_ClaseBaja', 
    'est_socio_2': 'cantidadHogares_ClaseMediaBaja', 
    'est_socio_3': 'cantidadHogares_ClaseMediaAlta', 
    'est_socio_4': 'cantidadHogares_claseAlta'
})
# Mostramos los primeros resultados
df_final
```

```python

# 1. Asegurarnos de tratar la columna como texto y quitar decimales si Pandas los agregó
# (A veces los números se leen como 10001.0, esto lo corrige a "10001")
df_final['ubica_geo_str'] = df_final['ubica_geo'].fillna(0).astype(int).astype(str)

# 2. Función para extraer el estado según la longitud
def extraer_estado(codigo):
    if len(codigo) == 4:
        return codigo[:1]  # Primer carácter
    elif len(codigo) == 5:
        return codigo[:2]  # Primeros dos caracteres
    else:
        # Por si hay algún código más corto o con ceros omitidos
        return  '0'

# 3. Aplicamos las reglas para crear las nuevas columnas
df_final['estado'] = df_final['ubica_geo_str'].apply(extraer_estado).astype(int)
df_final['municipio'] = df_final['ubica_geo_str'].apply(lambda x: x[-3:]).astype(int)# Siempre los últimos 3
df_final.drop(columns="ubica_geo_str", inplace=True)
df_final
```

```python
df_final=df_final[['cantidadHogares_ClaseBaja',
       'cantidadHogares_ClaseMediaBaja', 'cantidadHogares_ClaseMediaAlta',
       'cantidadHogares_claseAlta', 'EstimPagoPromedio', 'estado', 'municipio']]
```

```python
# 1. Renombrar columnas en df_mun_est
df_mun_est = df_mun_est.rename(columns={
    'CVE_ENT': 'estado',
    'CVE_MUN': 'municipio'
})

# 2. Merge con df_final
df_mun_estados = df_mun_est.merge(
    df_final,
    on=['estado', 'municipio'],
    how='left'
)

# 3. Calcular promedio municipal (por estado y municipio)
promedios = df_final.groupby(['estado','municipio']).mean().reset_index()


# 5. Rellenar valores faltantes con el promedio municipal
cols = [
    'cantidadHogares_ClaseBaja',
    'cantidadHogares_ClaseMediaBaja',
    'cantidadHogares_ClaseMediaAlta',
    'cantidadHogares_claseAlta',
    'EstimPagoPromedio'
]

# 1. Reemplazar ceros por NaN
df_mun_estados[cols] = df_mun_estados[cols].replace(0, np.nan)

# 2. Calcular promedios por estado
promedios_estado = df_mun_estados.groupby('estado')[cols].transform('mean')
promedios_estado.fillna(0, inplace=True)
# 3. Rellenar NaN con el promedio del mismo estado
for col in cols:
    df_mun_estados[col] = df_mun_estados[col].fillna(promedios_estado[col])


# 6. (Opcional) ordenar
df_mun_estados = df_mun_estados.sort_values(by=['estado','municipio']).reset_index(drop=True)
```
